```python
import ee
import geemap

# Initialize Earth Engine
ee.Initialize()

print("Earth Engine initialized successfully!")
```

```python
# Define Delhi region
delhi = ee.Geometry.Rectangle(
    [76.84, 28.40, 77.35, 28.90]
)

Map = geemap.Map()
Map.centerObject(delhi, 10)

Map.addLayer(delhi, {}, "Delhi")
Map
```

```python
sentinel_2024 = (
    ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED')
    .filterBounds(delhi)
    .filterDate('2024-01-01', '2024-12-31')
    .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE',10))
    .median()
)

rgb_vis = {
    'min':0,
    'max':3000,
    'bands':['B4','B3','B2']
}

Map = geemap.Map()
Map.centerObject(delhi,10)

Map.addLayer(
    sentinel_2024,
    rgb_vis,
    'Delhi 2024'
)

Map
```
